# GRID para distintas gráficas estadisticas

```python
def plot00(df, ax):
    sns.barplot(
        data=df,
        y=df.index,
        x='edad',
        ax=ax
    )
    ax.set_title("Distribución de edades")
    ax.grid(axis='x', alpha=0.3)

def plot01(df, ax):
    sns.barplot(
        data=df,
        y=df.index,
        x='ingresos',
        ax=ax
    )
    ax.set_title("Distribución de ingresos")
    ax.grid(axis='x', alpha=0.3)

fig, axs = plt.subplots(2, 2, figsize=(12, 6))

fig.suptitle("Análisis exploratorio")

plot00(df, axs[0, 0])
plot01(df, axs[0, 1])

axs[0, 0].grid()
axs[0, 1].grid()
axs[1, 0].set_title("Gráfica 3")
axs[1, 1].set_title("Gráfica 4")

plt.tight_layout()
plt.show()
```
